### TODO: Write a function that initializes omega and xi

Complete the function `initialize_constraints` so that it returns `omega` and `xi` constraints for the starting position of the robot. Any values that we do not yet know should be initialized with the value `0`. You may assume that our robot starts out in exactly the middle of the world with 100% confidence (no motion or measurement noise at this point). The inputs `N` time steps, `num_landmarks`, and `world_size` should give you all the information you need to construct intial constraints of the correct size and starting values.

*Depending on your approach you may choose to return one omega and one xi that hold all (x,y) positions *or* two of each (one for x values and one for y); choose whichever makes most sense to you!*

````python
def initialize_constraints(N, num_landmarks, world_size):
    ''' This function takes in a number of time steps N, number of landmarks, and a world_size,
        and returns initialized constraint matrices, omega and xi.'''
    
    # Two columns per pose (x, y) for N poses, plus two columns per landmark.
    size = 2 * (N + num_landmarks)
    omega = np.zeros((size, size))
    xi = np.zeros((size, 1))
    # Anchor the robot's starting pose at the centre of the world with full
    # confidence (a strength of 1 on the diagonal, the centre coordinate in xi).
    omega[0, 0] = 1
    omega[1, 1] = 1
    xi[0, 0] = world_size / 2
    xi[1, 0] = world_size / 2
    return omega, xi
    
````

#### A note on noise

Recall that `omega` holds the relative "strengths" or weights for each position variable, and you can update these weights by accessing the correct index in omega `omega[row][col]` and *adding/subtracting* `1.0/noise` where `noise` is measurement or motion noise. `Xi` holds actual position values, and so to update `xi` you'll do a similar addition process only using the actual value of a motion or measurement. So for a vector index `xi[row][0]` you will end up adding/subtracting one measurement or motion divided by their respective `noise`.

#### Updating with motion and measurements

With a 2D omega and xi structure as shown above (in earlier cells), you'll have to be mindful about how you update the values in these constraint matrices to account for motion and measurement constraints in the x and y directions. Recall that the solution to these matrices (which holds all values for robot poses `P` and landmark locations `L`) is the vector, `mu`, which can be computed at the end of the construction of omega and xi as the inverse of omega times xi: $\mu = \Omega^{-1}\xi$

**You may also choose to return the values of `omega` and `xi` if you want to visualize their final state!**

````python
## TODO: Complete the code to implement SLAM

## slam takes in 6 arguments and returns mu, 
## mu is the entire path traversed by a robot (all x,y poses) *and* all landmarks locations
def slam(data, N, num_landmarks, world_size, motion_noise, measurement_noise):
    
    omega, xi = initialize_constraints(N, num_landmarks, world_size)

    # Information weights — Cholesky factors of the noise variances.
    measurement_strength = 1.0 / measurement_noise
    motion_strength      = 1.0 / motion_noise

    for t, (measurements, motion) in enumerate(data):
        # Indices into omega/xi for the current pose and the next pose.
        x_t  = 2 * t           # x of pose t
        y_t  = x_t + 1         # y of pose t
        x_t1 = 2 * (t + 1)
        y_t1 = x_t1 + 1

        # Measurements relate the current pose to landmarks.
        for landmark_idx, dx, dy in measurements:
            l_x = 2 * (N + landmark_idx)
            l_y = l_x + 1
            for i, j, val in [
                (x_t, x_t,  measurement_strength),
                (l_x, l_x,  measurement_strength),
                (x_t, l_x, -measurement_strength),
                (l_x, x_t, -measurement_strength),
                (y_t, y_t,  measurement_strength),
                (l_y, l_y,  measurement_strength),
                (y_t, l_y, -measurement_strength),
                (l_y, y_t, -measurement_strength),
            ]:
                omega[i, j] += val
            xi[x_t, 0] += -dx * measurement_strength
            xi[l_x, 0] +=  dx * measurement_strength
            xi[y_t, 0] += -dy * measurement_strength
            xi[l_y, 0] +=  dy * measurement_strength

        # Motion relates pose t to pose t+1.
        dx, dy = motion
        for i, j, val in [
            (x_t,  x_t,   motion_strength),
            (x_t1, x_t1,  motion_strength),
            (x_t,  x_t1, -motion_strength),
            (x_t1, x_t,  -motion_strength),
            (y_t,  y_t,   motion_strength),
            (y_t1, y_t1,  motion_strength),
            (y_t,  y_t1, -motion_strength),
            (y_t1, y_t,  -motion_strength),
        ]:
            omega[i, j] += val
        xi[x_t,  0] += -dx * motion_strength
        xi[x_t1, 0] +=  dx * motion_strength
        xi[y_t,  0] += -dy * motion_strength
        xi[y_t1, 0] +=  dy * motion_strength

    # Closed-form best estimate.
    mu = np.linalg.inv(omega) @ xi
    return mu
````

## Helper functions

To check that your implementation of SLAM works for various inputs, we have provided two helper functions that will help display the estimated pose and landmark locations that your function has produced. First, given a result `mu` and number of time steps, `N`, we define a function that extracts the poses and landmarks locations and returns those as their own, separate lists. 

Then, we define a function that nicely print out these lists; both of these we will call, in the next step.


````python
# a helper function that creates a list of poses and of landmarks for ease of printing
# this only works for the suggested constraint architecture of interlaced x,y poses
def get_poses_landmarks(mu, N):
    # create a list of poses
    poses = []
    for i in range(N):
        poses.append((mu[2*i].item(), mu[2*i+1].item()))

    # create a list of landmarks
    landmarks = []
    for i in range(num_landmarks):
        landmarks.append((mu[2*(N+i)].item(), mu[2*(N+i)+1].item()))

    # return completed lists
    return poses, landmarks
````

````python
def print_all(poses, landmarks):
    print('\n')
    print('Estimated Poses:')
    for i in range(len(poses)):
        print('['+', '.join('%.3f'%p for p in poses[i])+']')
    print('\n')
    print('Estimated Landmarks:')
    for i in range(len(landmarks)):
        print('['+', '.join('%.3f'%l for l in landmarks[i])+']')
````

### Question: How far away is your final pose (as estimated by `slam`) compared to the *true* final pose? Why do you think these poses are different?

You can find the true value of the final pose in one of the first cells where `make_data` was called. You may also want to look at the true landmark locations and compare them to those that were estimated by `slam`. Ask yourself: what do you think would happen if we moved and sensed more (increased N)? Or if we had lower/higher noise parameters.

**Answer**:

For this particular run the *true* final robot pose printed by `make_data` was

> `Robot: [x=9.71640, y=67.02871]`

and the *estimated* final pose returned by my `slam` implementation was

> `Last pose: (10.0698, 64.8021)`

The Euclidean error is therefore roughly

`sqrt((10.0698 - 9.71640)^2 + (64.8021 - 67.02871)^2) ≈ 2.25`

units in a 100×100 world, i.e. about **2.3 % of the world side length** — a very accurate estimate. The five landmark estimates are similarly close to their true coordinates (the largest landmark error is only ~1.1 units).

**Why isn't the estimate exactly right?**

* `make_data` uses two independent stochastic noise sources — `motion_noise=2.0` (every `r.move(dx, dy)` lands at the requested point *plus* zero-mean Gaussian noise) and `measurement_noise=2.0` (every `r.sense()` reading of `(landmark - robot)` is also corrupted by Gaussian noise). My `slam` only knows the *noisy* motions and *noisy* measurements that `make_data` recorded; it has no direct access to the true motion or sensor readings.
* Graph-SLAM does the *best linear unbiased* thing it can with that data — it solves `μ = Ω⁻¹ ξ`, where each constraint is weighted by the inverse of the corresponding noise variance (`1 / motion_noise` and `1 / measurement_noise`). Because both noise sources are non-zero, the resulting estimate has finite variance. So a residual error of a couple of units is exactly what you'd expect from this noise level over ~20 motion steps.
* In addition, errors compound along the trajectory: the further the robot has moved from the anchored initial pose at `(50, 50)`, the more the accumulated motion noise bends the estimate.

**What would change?**

* **Increasing `N` (more motion + sensor steps)** generally *reduces* the per-pose error, because every additional sensor reading introduces another constraint into `Ω` that locks the landmark and pose estimates more tightly together. With many measurements per landmark the estimate of each landmark approaches its true value, and that in turn pulls every connected pose closer to its true value.
* **Lower noise (`motion_noise` and/or `measurement_noise` → 0)** approaches the noiseless limit where `μ` becomes the exact ground truth (because `1 / noise → ∞` makes each constraint hard rather than soft). **Higher noise** has the opposite effect: each constraint is downweighted, the inverse problem becomes more under-determined, and the recovered trajectory and landmark estimates drift further from the truth.
